```python
import numpy as np
import pandas as pd
import difflib  # Compare the input with the dataset
from sklearn. feature_extraction.text import TfidfVectorizer # Convert the textual features into numbers
from sklearn.metrics.pairwise import cosine_similarity # Identify similar movies 

df= pd.read_csv("movies.csv")
selected_features=['genres','keywords','tagline','cast','director']

for features in selected_features:
    df[features]= df[features].fillna('')
    
combined_features = df['genres'] +' '+ df['keywords'] +' '+ df['tagline'] +' '+ df['cast'] +' ' +df['director'] 

# Converting the text "combined_features" into feature vectors
vectorizer =TfidfVectorizer()
feature_vectors= vectorizer.fit_transform(combined_features )

# Fetching the similarity score using the Cosine Similarity Algorithm
similarity =cosine_similarity(feature_vectors)# Generate a square matrix comparing each movie to all the other movies

# Taking the name of a movie from the user as input
movie_name = input ('Enter the name of your favourite movie :')

#Creating a list of all movies in the given dataset
list_of_all_movies= df['title'].tolist()

# Find a close match for the given movie in the given dataset
find_close_match= difflib.get_close_matches(movie_name,list_of_all_movies )

close_match= find_close_match[0] # Taking the first match as the most reliable 

# Finding the index of the movie using the title
index_of_the_movie= df[df.title==close_match ]['index'].values[0]

# Getting a list of similar movies
similarity_score= list(enumerate(similarity[index_of_the_movie])) # first value represents the index of the movie and the second value represents the similarity score

# Sorting the similarity_score list in descending order 
sorted_similarity_score= sorted(similarity_score, key = lambda x:x[1], reverse= True)#first value represent index of the movie and the second value represent the similarity score

# Print the name of the similar movies
print ('Movies suggested for you:\n')
i=1
for score in sorted_similarity_score:
    index= score[0]
    title_from_index=df[df.index==index]['title'].values[0] # .values [0] extracts the name of the movie as text from the pandas series
    if (i<11): # Suggesting top 10 movies 
        print (i, ')',title_from_index)
        i=i+1
        
```
